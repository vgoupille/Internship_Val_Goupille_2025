```python
%%R
# Install and load required packages
if (!require("Seurat", quietly = TRUE)) {
    install.packages("Seurat", repos = "https://cloud.r-project.org/")
}
if (!require("ggplot2", quietly = TRUE)) {
    install.packages("ggplot2", repos = "https://cloud.r-project.org/")
}
if (!require("dplyr", quietly = TRUE)) {
    install.packages("dplyr", repos = "https://cloud.r-project.org/")
}

# Load libraries
library(Seurat)
library(ggplot2)
library(dplyr)

cat("✅ Packages loaded successfully\n")
cat("Seurat version:", as.character(packageVersion("Seurat")), "\n")
```

```python
%%R
# Define path to RDS file
rds_path <- "data/anndata_objects_with_complete_metadata_filtered/GeneFull_raw_unique_mult_with_metadata_grouped_filtered_100UMI_3c_SeuratV5.rds"

# Check if file exists
if (!file.exists(rds_path)) {
    stop("❌ RDS file not found at: ", rds_path)
}

cat("📂 Loading RDS file from:", rds_path, "\n")

# Load the Seurat object
seurat_obj <- readRDS(rds_path)

cat("✅ RDS file loaded successfully\n")
```

```python
%%R
# Display basic information about the Seurat object
cat("📊 SEURAT OBJECT SUMMARY\n")
cat(rep("=", 50), "\n")
print(seurat_obj)

cat("\n📊 OBJECT DIMENSIONS\n")
cat("Number of cells:", ncol(seurat_obj), "\n")
cat("Number of genes:", nrow(seurat_obj), "\n")

cat("\n🧬 AVAILABLE ASSAYS\n")
print(names(seurat_obj@assays))

cat("\n📋 OBJECT CLASS\n")
print(class(seurat_obj))
```

```python
%%R
# Display metadata information
cat("📊 METADATA OVERVIEW\n")
cat(rep("=", 50), "\n")

cat("Number of metadata columns:", ncol(seurat_obj@meta.data), "\n")
cat("\nMetadata columns:\n")
print(colnames(seurat_obj@meta.data))

cat("\n📋 First few rows of metadata:\n")
print(head(seurat_obj@meta.data, 3))

# Check important columns
important_cols <- c("CultureMedium", "OD", "RepBio", "RepTech", "total_counts", "n_genes")

cat("\n🔍 CHECKING IMPORTANT COLUMNS\n")
for (col in important_cols) {
    if (col %in% colnames(seurat_obj@meta.data)) {
        cat("✅", col, "- Available\n")
        if (col %in% c("CultureMedium", "OD", "RepBio", "RepTech")) {
            unique_vals <- unique(seurat_obj@meta.data[[col]])
            cat("   Unique values:", paste(unique_vals, collapse = ", "), "\n")
        }
    } else {
        cat("❌", col, "- Not found\n")
    }
}
```

```python
%%R
# Final validation and summary
cat("\n", rep("=", 80), "\n")
cat("📊 SEURAT RDS FILE VALIDATION SUMMARY\n")
cat(rep("=", 80), "\n")

cat("✅ RDS file loaded successfully\n")
cat("✅ Seurat object structure valid\n")
cat("✅ Expression matrix accessible\n")
cat("✅ Metadata preserved\n")

cat("\n📊 Object Statistics:\n")
cat("   Cells:", ncol(seurat_obj), "\n")
cat("   Genes:", nrow(seurat_obj), "\n")
cat("   Metadata columns:", ncol(seurat_obj@meta.data), "\n")
cat("   Assays:", length(seurat_obj@assays), "\n")
cat("   Seurat version:", as.character(packageVersion("Seurat")), "\n")

# Test basic functionality
cat("\n🧪 Testing basic functions:\n")
test_subset <- seurat_obj[1:100, 1:10]
cat("✅ Subsetting works - subset dimensions:", dim(test_subset), "\n")

counts_matrix <- GetAssayData(seurat_obj, assay = "RNA", slot = "counts")
cat("✅ Matrix access works - matrix class:", class(counts_matrix), "\n")

cat("\n🎉 AnnData to Seurat V5 conversion SUCCESSFUL!\n")
cat("✅ Object ready for downstream Seurat analysis\n")

cat("\n💡 Next steps:\n")
cat("   - Normalization: NormalizeData(seurat_obj)\n")
cat("   - Variable features: FindVariableFeatures(seurat_obj)\n")
cat("   - Scaling: ScaleData(seurat_obj)\n")
cat("   - PCA: RunPCA(seurat_obj)\n")
cat("   - Clustering: FindClusters(seurat_obj)\n")
cat("   - UMAP: RunUMAP(seurat_obj)\n")
```
