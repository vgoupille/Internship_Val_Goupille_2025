## Add genes annotations 

```r
# Fast function to handle duplicate symbols in annotation file and modify locus_tag
handle_duplicate_symbols <- function(tsv_file) {
    # Read the TSV file
    annotation_data <- read.delim(tsv_file, stringsAsFactors = FALSE, sep = "\t")

    # Create a copy of original symbols
    annotation_data$symbol_old <- annotation_data$symbol

    # Create a copy of original locus_tag and modify it
    if ("locus_tag" %in% colnames(annotation_data)) {
        annotation_data$locus_tag_old <- annotation_data$locus_tag
        # Replace underscores with hyphens in locus_tag
        annotation_data$locus_tag <- gsub("_", "-", annotation_data$locus_tag)
        cat("Modified locus_tag: replaced underscores with hyphens\n")
    } else {
        cat("Warning: locus_tag column not found in annotation file\n")
    }

    # Find duplicates in symbol column
    symbol_counts <- table(annotation_data$symbol)
    duplicates <- names(symbol_counts[symbol_counts > 1])

    cat("Found", length(duplicates), "duplicate symbols\n")

    if (length(duplicates) > 0) {
        # Process each duplicate symbol
        for (dup_symbol in duplicates) {
            if (dup_symbol != "") { # Skip empty symbols
                # Find rows with this duplicate symbol
                dup_indices <- which(annotation_data$symbol == dup_symbol)

                # Rename symbols with suffixes (_1, _2, etc.)
                for (i in seq_along(dup_indices)) {
                    annotation_data$symbol[dup_indices[i]] <- paste0(dup_symbol, "_", i)
                }

                cat("Renamed", length(dup_indices), "instances of '", dup_symbol, "'\n")
            }
        }
    }

    # Add column indicating if symbol was renamed
    annotation_data$symbol_renamed <- annotation_data$symbol != annotation_data$symbol_old

    annotation_data$locus_tag_a <- annotation_data$locus_tag


    #  add column named final_gene_name which is the locus_tag but if symbol exists for a row take the symbol

    annotation_data <- annotation_data %>%
        mutate(final_gene_name = case_when(
            !is.na(symbol) & symbol != "" ~ symbol,
            TRUE ~ locus_tag
        ))

    # Create new filename with suffix
    file_path <- file.path(dirname(tsv_file), paste0(tools::file_path_sans_ext(basename(tsv_file)), "_no_duplicates.tsv"))

    # Write the modified data to new file
    write.table(annotation_data, file_path, sep = "\t", row.names = FALSE, quote = FALSE)

    cat("New annotation file saved as:", file_path, "\n")
    cat("Total rows renamed:", sum(annotation_data$symbol_renamed), "\n")

    return(annotation_data)
}
```

```r
# Apply the function to handle duplicates
annotation_data_modified <- handle_duplicate_symbols(annotation_file)
```
