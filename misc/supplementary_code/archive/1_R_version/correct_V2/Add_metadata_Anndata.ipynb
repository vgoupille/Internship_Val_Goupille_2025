# Import the Anndata

```python
# Import necessary libraries"anndata"
import anndata as ad
import pandas as pd
import numpy as np
```

```python
# Import the Anndata object
adata = ad.read_h5ad(
    "jupyter/correct/correct_V2/ScObjects/SeuratObject_Val_Filtered_CDS_5cells.h5ad"
)
```

```python
adata.var.head()
```

```python
# Import the tsv file contain annotation of genes
gene_annotation = pd.read_csv(
    "data/raw_data/genome_annotation/other/extracted_features_no_duplicates.tsv",
    sep="\t",
)
```

```python
gene_annotation.head()
```

```python
# add the gene annotation to the adata object using the column (final_gene_name) of the gene_annotation dataframe and the index of adata.var

# transfere toute les colonnes du tsv en utilisant la colonne final_gene_name comme index qui correspond à l'index de adata.var
adata.var = adata.var.join(gene_annotation.set_index("final_gene_name"), how="left")
```

```python
adata.var.head()
```

```python
for col in adata.var.columns:
    unique_vals = adata.var[col].unique()
    print(f"🧬 Column: {col}")
    print(f"🔢 Number of unique values: {len(unique_vals)}")
    print(f"📋 Unique values: {unique_vals}\n")
```

```python
# save the adata object with the new gene annotation
adata.write_h5ad(
    "jupyter/correct/correct_V2/ScObjects/SeuratObject_Val_Filtered_CDS_5cells_with_gene_annotation.h5ad"
)

print("✅ Gene annotation added and saved successfully!")
```
